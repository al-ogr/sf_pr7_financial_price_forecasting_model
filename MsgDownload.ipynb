## Загрузка и первичный обзор данных

```python
df_msgs = pd.read_json('./data/markettwits_posts_2017-01-01_2026-07-27/posts_markettwits_2017-01-01_2026-07-27.json')
df_msgs.info()
```

```python
df_msgs.describe(include='all')
```

Датасет содержит следующие столбцы:
- **id** - идентификатор сообщения в Телеграме (*подлежит удалению*);
- **date_utc** - время публикации сообщения;
- **message** - текст сообщения;
- **views** - количество просмотров (изменяется во времени, актуальное значение можно получить только в моменте, значение можно использовать при построении системы, работающей в режиме реального времени, для определения степени распространения информации и использовать для улучшения работы модели в последующих итерациях разработки, *подлежит удалению*);
- **forwards** - количество ретрансляций сообщения (комментарий тот же, что и для views, *подлежит удалению*);
- **replies** - все значения пустые, для канала не актуально (*подлежит удалению*);
- **has_media** - наличие прикрепленной мультимедийной информации (*подлежит удалению*, мультимедиа на данной итерации не обрабатывается);
- **link** - ссылка на сообщение (*подлежит удалению*).

Выгруженные данные содержат сообщения канала с 8:38 9 ноября 2017 г. по 15:56 27 июля 2026 г., 370163 записей сообщений.
Произведем первичную очистку датасета от отмеченных столбцов, а также от записей без текста сообщений.

```python
drop_columns = ['id', 'views', 'forwards', 'replies', 'has_media', 'link']
df_msgs_clr = df_msgs.drop(columns=drop_columns)
df_msgs_clr.info()
```

```python
df_msgs_clr = df_msgs_clr[df_msgs_clr['message'] != '']
df_msgs_clr
```

Предварительно очищенный датасет сообщений канала содержит 366965 сообщений. Для дальнейшей обработки и анализа данных выгружаем датасет в файл.

```python
df_msgs_clr.to_csv('./data/tg_messages.csv', index=False)
```
